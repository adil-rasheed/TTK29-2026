# Module 22 — Physics-informed neural networks (PINNs)

**Start here.** Click a link to open that notebook. Work through them **in the order below**; it follows the lecture slides section by section.

**Before you start**
- Each notebook begins with a **Parameters (edit these)** cell. Run it once with the defaults first (*Run → Run All Cells*). The default run reproduces the numbers on the slides, and the asserts at the end check them. Then change the parameters and run it again.
- Each notebook is designed to finish in about 2 minutes on a laptop CPU. A GPU is not needed.
- Open the `.ipynb` files only. The `.py` file with the same name is the same notebook stored as plain text (see the note at the bottom).

## Optional: seed sweeps

`pinn_heat_pde` and `pinn_wind_pde` each have a `RUN_SEED_SWEEP = False` switch in their parameter cell. Set it to `True` to repeat the experiment over several random seeds: about 6 and 5 minutes respectively. The saved results are already in `data/`, so the notebooks read them by default. The sweep shows how much a PINN result changes between seeds.

## Note: why is there a `.py` next to every notebook?

Each `foo.py` next to `foo.ipynb` is the **same notebook**. The two are paired with [jupytext](https://jupytext.readthedocs.io), and the `.py` holds the same cells as plain text (`# %%` marks a cell). It is kept for version control because it gives readable diffs in git, while an `.ipynb` is JSON and also stores outputs. Work in the `.ipynb`.
