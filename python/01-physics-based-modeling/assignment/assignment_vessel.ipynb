## Task 1 — Zig-zag overshoot

Run a zig-zag maneuver and compute the **heading overshoot**: the maximum $|\psi|$ reached beyond the switching angle `psi_switch=0.35` rad.

```python
psi_switch = 0.35
### BEGIN SOLUTION
t, Z = vessel.simulate(vessel.zigzag(psi_switch=psi_switch), t_end=80.0, dt=0.02)
overshoot = np.max(np.abs(Z[:,2])) - psi_switch
### END SOLUTION
print('overshoot [rad]:', round(float(overshoot), 3))
assert overshoot > 0
```

## Task 2 — Sensitivity of turning radius to added mass

Sweep the sway added mass `Y_vdot` over `[-50,-30,-10]` for a turning circle and report the approximate turning radius (half the path's y-extent).

```python
radii = {}
### BEGIN SOLUTION
for Yv in [-50,-30,-10]:
    p = vessel.VesselParams(Y_vdot=Yv)
    _, Z = vessel.simulate(vessel.turning_circle(), t_end=60.0, dt=0.02, p=p)
    radii[Yv] = 0.5*(Z[:,1].max()-Z[:,1].min())
### END SOLUTION
print({k: round(float(v),2) for k,v in radii.items()})
assert len(radii) == 3
```
