## Task 1 — Cavity Reynolds-number sweep

Run the lid-driven cavity at `Re in [100, 400]` and report the maximum speed for each. (Keep the grid small.)

```python
maxspeed = {}
### BEGIN SOLUTION
for Re in [100, 400]:
    x,y,u,v,p = ns.lid_driven_cavity(nx=41, ny=41, nt=400, Re=Re)
    maxspeed[Re] = float(np.sqrt(u**2+v**2).max())
### END SOLUTION
print({k: round(v,3) for k,v in maxspeed.items()})
assert set(maxspeed) == {100,400}
```

## Task 2 — Vortex merging vs initial separation

Run the vortex merger for separations `sep in [0.4, 0.8, 1.2]` and decide, for each, whether the two cores have merged into a single peak by the end (hint: count local maxima above half the peak vorticity along the mid-line).

```python
merged = {}
### BEGIN SOLUTION
for sep in [0.4, 0.8, 1.2]:
    X,Y,w = ns.vortex_merger(n=96, nt=2000, Re=2000.0, dt=1e-3, sep=sep)
    line = w[:, w.shape[1]//2]
    thr = 0.5*line.max()
    peaks = np.sum((line[1:-1] > thr) & (line[1:-1] > line[:-2]) & (line[1:-1] > line[2:]))
    merged[sep] = (peaks <= 1)
### END SOLUTION
print(merged)
assert len(merged) == 3
```
