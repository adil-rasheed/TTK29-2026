## 1. Compressed sensing — do these in order

| # | Notebook | Slide section | What you learn |
|---|----------|---------------|----------------|
| 1 | [cs_sparsity](notebooks/cs_sparsity.ipynb) | Sparsity: the premise, measured | Sparsity is a property of a signal *and* a basis: a two-tone signal is dense in time and 2-sparse in the DCT; a real photograph is compressible (1%, 5%, 20% of its DCT coefficients). The sorted-magnitude test that tells you whether compressed sensing can work at all. **Do this first.** |
| 2 | [cs_illustrations](notebooks/cs_illustrations.ipynb) | Sparsity; The recovery problem | The whole subject as matrix pictures: $\boldsymbol{y}=\mathbf{C}\boldsymbol{x}$, $\boldsymbol{x}=\boldsymbol{\Psi}\boldsymbol{s}$, $\boldsymbol{y}=\boldsymbol{\Phi}\boldsymbol{s}$. | 3 | [cs_regularization](notebooks/cs_regularization.ipynb) | The recovery problem | Why $\ell_1$ gives exact zeros and $\ell_2$ does not: norm balls, where the loss first touches the ball, soft-thresholding in closed form, and how often the answer is sparse (counted over 500 random problems). |
| 4 | [cs_phase_transition](notebooks/cs_phase_transition.ipynb) | The recovery problem | How many measurements you need: the success rate over a grid of sparsity $k$ and measurements $m$, and the fitted law $m\approx C\,k\log(n/k)$. |
| 5 | [cs_recovery_fista](notebooks/cs_recovery_fista.ipynb) | Algorithms: solving the $\ell_1$ problem | ISTA, FISTA and OMP written from scratch and checked against scikit-learn; why momentum helps; debiasing; recovering two tones from 128 random samples, far below the Nyquist rate. |
| 6 | [cs_uniform_vs_random](notebooks/cs_uniform_vs_random.ipynb) | Algorithms: solving the $\ell_1$ problem | Same number of samples, two methods: uniform samples with sinc interpolation against random samples with basis pursuit, swept from 8 to 232 samples. |
| 7 | [cs_portrait_fista](notebooks/cs_portrait_fista.ipynb) | Algorithms; What compressed sensing is for | Reconstructing the portrait from random pixels with FISTA, choosing $\lambda$, an honest comparison with linear interpolation (which wins on clean pixels), and the case where compressed sensing wins: a single-pixel camera with a noisy detector. |

## Note: why is there a `.py` next to every notebook?

Each `foo.py` next to `foo.ipynb` is the **same notebook**. The two are paired with [jupytext](https://jupytext.readthedocs.io), and the `.py` holds the same cells as plain text (`# %%` marks a cell). It is kept for version control because it gives readable diffs in git, while an `.ipynb` is JSON and also stores outputs. Work in the `.ipynb`.
